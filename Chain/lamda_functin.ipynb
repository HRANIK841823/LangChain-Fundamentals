### lambda arguments : expression

```python
# lambda function
square=lambda number: number * number

square(5)
```

### Conditional Lambda


```python
def is_adult(age):
    return age>=18

is_adult(18)
```

```python
is_adult=lambda age : age >=18
#function_name=lambda varibale: condition
is_adult(17)
```

```python
x={"question":"What is 2 +3 ?"}
```

```python
def check_question(x):
    for ch in x['question']:
        if ch.isdigit():
            return True
    return False




print(check_question(x))
```

```python
#smaller version
def check_question(x):
   return [ch.isdigit() for ch in x['question']]
```

```python
check_question(x)
```

```python
#smaller version
def check_question(x):
   return any(ch.isdigit() for ch in x['question'])
```

```python
check_question(x)
```

### using lambda function

```python
s=lambda x : any(ch.isdigit() for ch in x['question'])
```
